```python
# 4-1 

import numpy as np 
import matplotlib.pyplot as plt 

dt = 0.1
t_end = 50
t = np.arange(0, t_end + dt, dt)

N = len(t) # 501 
print(N)


F = np.array([[1.0, dt],
              [0.0, 1.0]])

x = np.array([[1.0],                      # (2,1) 열벡터
              [3.0]])                     # → F @ x 가 (2,1)로 깔끔


x_history  = np.zeros((N,2))
x_history[0] = x.flatten()

print(x)


print(x.flatten())  # 여기서 이렇게 해둬도 for문에서 원래 형태를 유지함 


for k in range(1,N):
    x = F @ x

    x_history[k] = x.flatten()
    
    
    
fig, axes = plt.subplots(2,1,figsize=(8,6))

axes[0].plot(t,x_history[:,0],'b-')
axes[0].set_title("Position")
axes[0].set_xlabel("t")
axes[0].set_ylabel("p [m]")
axes[0].grid()


axes[1].plot(t,x_history[:,1],'r-')
axes[1].set_title("sim Velocity")
axes[1].set_xlabel("t")
axes[1].set_ylabel("v [m/s]")
axes[1].grid()

plt.tight_layout()

plt.show()




```

```python
# 4-2

import numpy as np 
H = np.array([1.0, 0.0])

np.random.seed(42)
sigma = 1
w = np.random.normal(0,sigma, N)

y_history = H @ x_history[:,:].transpose() + w

y_history = y_history.flatten()

# 5️⃣ 결과 시각화
#    원본(파란 실선) vs 측정값(빨간 점선) 겹쳐 그리기
# =========================================
plt.figure(figsize=(9, 5))
plt.plot(t, x_history[:, 0], 'b-',  label='original (true)', linewidth=1.5)
plt.plot(t, y_history,        'r--', label='noise (measurement)', linewidth=0.8)

plt.title('Measurement Position from original and Measurement Position from noise')
plt.xlabel('time [s]')
plt.ylabel('position [m]')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()


# =========================================
# 6️⃣ 노이즈 통계 확인 (검증용)
#    - 평균이 0에 가까운지
#    - 표준편차가 1에 가까운지
# =========================================
print(f"노이즈 평균   : {np.mean(w):.4f} (이론: 0)")
print(f"노이즈 표준편차: {np.std(w):.4f}  (이론: {sigma})")
print(f"측정값 개수   : {len(y_history)}")
```
