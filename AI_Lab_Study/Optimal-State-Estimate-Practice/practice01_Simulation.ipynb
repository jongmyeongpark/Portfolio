Practice 1 풀이 가이드
핵심 아이디어
상태벡터 [r, v, a]를 매 타임스텝마다 행렬 곱으로 업데이트하는 것이 전부예요.

Step 1. 파라미터 설정
dt = 0.1,
g = -9.80665,
v0 = 40,
r0 = 100

Step 2. 상태전이 행렬 F 만들기
PDF의 3×3 행렬을 numpy 배열로 표현하세요.
F = [[1, dt, dt²/2],
     [0,  1,    dt],
     [0,  0,     1]]

Step 3. 초기 상태벡터 설정
x = [r0, v0, g]  # (3,1) 벡터

Step 4. 시뮬레이션 루프

t = 0부터 t = 10까지, dt = 0.1이니까 총 몇 스텝인지 먼저 계산해보세요.
매 루프마다 x = F @ x 로 상태 업데이트
각 스텝의 r, v, a를 리스트에 저장

Step 5. 플롯

matplotlib 사용
subplot 3개로 r, v, a를 각각 그리기
x축은 시간, y축은 각 상태값


막힐 수 있는 포인트 힌트


numpy.dot() 또는 @ 연산자로 행렬 곱
시간 배열은 np.arange(0, 10+dt, dt) 활용
r이 0 이하가 되면 (땅에 닿으면) 어떻게 처리할지 생각해보세요

```python
# ============================================================
# 단계 1: 라이브러리 임포트
# ============================================================
import numpy as np 
import matplotlib.pyplot as plt 

# ============================================================
# 단계 2: 파라미터 및 초기값 설정
# ============================================================
dt = 0.1 
g = -0.980665

r0 = 100.0
v0 = 0.0
a0 = g

t_end = 10.0 
N = int(t_end/dt)

# ============================================================
# 단계 3: 상태천이행렬 F 정의
# ============================================================
F = np.array([
    [1, dt, 0.5*dt**2],
    [0, 1, dt],
    [0, 0, 1]
])

# ============================================================
# 단계 4: 초기 상태벡터 설정 및 저장 배열 초기화
# ============================================================
x = np.array([r0, v0, a0]) # 초기 상태 벡터

time_arr = np.zeros(N+1)
r_arr = np.zeros(N+1)

v_arr = np.zeros(N+1)
a_arr = np.zeros(N+1)

time_arr[0] = 0.0
r_arr[0] = x[0]
v_arr[0] = x[1]
a_arr[0] = x[2]

# ============================================================
# 단계 5: 시뮬레이션 루프 (x_{t+1} = F @ x_t)
# ============================================================
for i in range(1, N+1): # start 이상 stop 미만
    x = F @ x # x next 로 하면 나중에 x = x_next 다시 해줘야함  그리고 이거 계산해줘야 해서 0을 앞으로 빼야됨 
    time_arr[i] = i*dt
    r_arr[i] = x[0]
    v_arr[i] = x[1]
    a_arr[i] = x[2] 
    

fig, axes = plt.subplots(3, 1, figsize=(10, 8)) # subplots() 여러 개의 그래프(axes)를 한 figure 안에 배치할 때 씀 
# fig 는 도화지 전체, axes 는 각각의 그래프 영역 
fig.suptitle('Simulation of Free Fall under Gravity') # supertitle

axes[0].plot(time_arr, r_arr, color='steelblue')
axes[0].set_ylabel("Altitude (m)")
axes[0].set_xlabel("Time (s)")
axes[0].grid()

axes[1].plot(time_arr, v_arr, color='seagreen')
axes[1].set_ylabel("Velocity v (m/s)")
axes[1].set_xlabel("Time (s)")
axes[1].grid()

axes[2].plot(time_arr, a_arr, color='tomato') # (x,y,color)
axes[2].set_ylabel("Acceleration a (m/s²)") # y축 라벨 
axes[2].set_xlabel("Time (s)") # x축 라벨 
axes[2].grid(True) # 격자 그리드 

plt.tight_layout() # 줄 간격 정리 
plt.show() # 출력
```

```python
# 1단 계 라이브러리 임포트 
import numpy as np 
import matplotlib.pyplot as plt 

# 2단계 파라미터 및 초기값 설정 
dt = 0.1
g = -9.80665 

r0 = 100
v0 = 40
a0 = g 

t_end = 10
N = int(t_end/dt)

# 3단계 상태전이행렬 F 정의 
F = np.array([
    [1, 1, 0.5*dt**2],
    [0, 1, dt],
    [0, 0, 1]
])

# 4단계 초기 상태 벡터 설정 및 저장 배열 초기화 
x = np.array([r0, v0, a0]) # 초기 상태 초기화 

time_arr = np.zeros(N+1)
r_arr = np.zeros(N+1)
v_arr = np.zeros(N+1)
a_arr = np.zeros(N+1)

time_arr[0] = 0.0
r_arr[0] = x[0]
v_arr[0] = x[1]
a_arr[0] = x[2]

for i in range(1,N+1):
    x = F @ x 
    time_arr[i] = i*dt
    r_arr[i] = x[0]
    v_arr[i] = x[1]
    a_arr[i] = x[2]
    
fig, axes = plt.subplots(3, 1, figsize=(10,8))
fig.suptitle("simulation of free fall under gravity")

axes[0].plot(time_arr,r_arr,color="blue")
axes[0].set_ylabel("Altitude (m)")
axes[0].set_xlabel("time (m)")
axes[0].grid()

axes[1].plot(time_arr,v_arr,color="blue")
axes[1].set_ylabel("velocity (m)")
axes[1].set_xlabel("time (m)")
axes[1].grid()


axes[2].plot(time_arr,a_arr,color="blue")
axes[2].set_ylabel("acceleration (m)")
axes[2].set_xlabel("time (m)")
axes[2].grid()

plt.tight_layout()
plt.show()
```

# 🔵 Practice 1 문제 풀이 (Simulation)

## ■ 목표
자유 낙하 물체의 상태(고도 $r_t$, 속도 $v_t$, 가속도 $a_t$)를 시간에 따라 시뮬레이션

---

## ■ 상태 정의

상태 벡터:
$$
x_t = 
\begin{bmatrix}
r_t \\
v_t \\
a_t
\end{bmatrix}
$$

---

## ■ 상태 방정식 (State Equation)

$$
r_{t+1} = r_t + v_t \Delta t + \frac{1}{2} a_t \Delta t^2
$$

$$
v_{t+1} = v_t + a_t \Delta t
$$

$$
a_{t+1} = a_t
$$

---

## ■ 조건

- 샘플링 시간: $\Delta t = 0.1$
- 중력 가속도: $a = -9.80665 \, m/s^2$
- 초기 속도: $v_0 = 40 \, m/s$
- 초기 고도: $r_0 = 100 \, m$
- 시뮬레이션 시간: $t = 0 \sim 10$

---

## ■ 풀이 절차

### 1️⃣ 시간 배열 생성
- 시작: $t = 0$
- 끝: $t = 10$
- 간격: $\Delta t = 0.1$
- 총 step 수 계산

---

### 2️⃣ 상태 저장 배열 생성
- $r$ (고도)
- $v$ (속도)
- $a$ (가속도)

또는 하나의 상태 행렬로 구성:
$$
X = [r, v, a]
$$

---

### 3️⃣ 초기값 설정
- $r[0] = 100$
- $v[0] = 40$
- $a[0] = -9.80665$

---

### 4️⃣ 상태 업데이트 (for loop)

각 시간 step마다 다음을 반복:

$$
r_{t+1} = r_t + v_t \Delta t + \frac{1}{2} a_t \Delta t^2
$$

$$
v_{t+1} = v_t + a_t \Delta t
$$

$$
a_{t+1} = a_t
$$

---

### 5️⃣ 결과 검증

- $r_t$는 감소해야 함 (물체 낙하)
- $v_t$는 점점 감소 (중력 영향)
- $a_t$는 일정 (중력 가속도 유지)

---

### 6️⃣ 결과 시각화 (Plot)

- x축: 시간 $t$
- y축:
  - $r_t$ (고도)
  - $v_t$ (속도)
  - $a_t$ (가속도)

---

## ■ 핵심 요약

- 이 문제는 **이산 상태공간 모델 기반 시뮬레이션**
- 이전 상태를 이용해 다음 상태를 반복적으로 계산
- Kalman Filter의 **Prediction 단계와 동일한 구조**

---

## ■ 한줄 정리

👉 "초기 상태에서 시작하여 상태 방정식을 반복 적용하여 시스템의 시간에 따른 변화를 계산하는 문제"


## 사고 정리 
- 시간 배열 만든다

- 상태 배열 만든다

- 초기값 넣는다

- for문 돌린다

- 상태 업데이트 식 적용한다

- 결과 저장한다

- 그래프 그린다

```python
# =========================================
# 1️⃣ 시간 배열 생성
# =========================================
import numpy as np
import matplotlib.pyplot as plt

dt = 0.1
t = np.arange(0, 10 + dt, dt)   # 0 ~ 10 # np.arange(start, end, step) 끝 포함 안됨, np 배열로 생성됨

N = len(t) # N = 101

# =========================================
# 2️⃣ 상태 저장 배열 생성 배열의 크기는 시간축 크기 만큼 
# =========================================
r = np.zeros(N)   # altitude
v = np.zeros(N)   # velocity
a = np.zeros(N)   # acceleration

# =========================================
# 3️⃣ 초기값 설정
# =========================================
r[0] = 100.0
v[0] = 40.0
a[0] = -9.80665

# =========================================
# 4️⃣ 상태 업데이트 (for loop) 이게 prediction 이자 propagation
# =========================================
for i in range(N - 1): # range(start, end, step) end 포함 안됨, 0 부터 100까지 
    r[i+1] = r[i] + v[i]*dt + 0.5*a[i]*(dt**2)
    v[i+1] = v[i] + a[i]*dt
    a[i+1] = a[i]   # constant acceleration

# =========================================
# 5️⃣ 결과 확인 (간단 체크)
# =========================================
print("최종 고도:", r[-1])
print("최종 속도:", v[-1])
print("가속도 (항상 동일):", a[0])

# =========================================
# 6️⃣ Plot
# =========================================

### plot 코드를 줄일 수 있는 방법 ### 
# fig, axes = plt.subplots(3, 1, figsize=(10, 8))

# data = [r, v, a]
# labels = ['Altitude (m)', 'Velocity (m/s)', 'Acceleration (m/s^2)']

# for i in range(3):
#     axes[i].plot(t, data[i])
#     axes[i].set_ylabel(labels[i])
#     axes[i].grid()

# axes[0].set_title('Free Fall Simulation')
# axes[2].set_xlabel('Time (s)')

# plt.tight_layout()
# plt.show()

plt.figure(figsize=(10, 8)) # 전체 그래프 판 생성 가로 10 세로 8

# 1️⃣ 고도
plt.subplot(3,1,1) # 3행 1열로 나누고 거기서 1번째 칸 
plt.title('Free Fall Simulation')
plt.plot(t, r) # 시간 배열과 r 배열이 x,y축에 각각 대응 되어서 들어감 
plt.ylabel('Altitude (m)')
plt.grid()

# 2️⃣ 속도
plt.subplot(3,1,2)
plt.plot(t, v)
plt.ylabel('Velocity (m/s)')
plt.grid()

# 3️⃣ 가속도
plt.subplot(3,1,3)
plt.plot(t, a)
plt.xlabel('Time (s)')
plt.ylabel('Acceleration (m/s^2)')
plt.grid()

plt.tight_layout() # 그래프 간격 자동 조정, 웬만하면 키는게 좋음 
plt.show()
```

# 🔵 Practice 2 문제 풀이 (Range Sensor Simulation)

## ■ 목표
Practice 1에서 생성한 실제 고도 $r_t$에 센서 노이즈를 추가하여 측정값 $z_t$를 생성

---

## ■ 측정 모델 (Measurement Equation)

$$
z_t = r_t + w_t
$$

- $w_t \sim \mathcal{N}(0, \sigma^2)$
- $\sigma = 0.1 \, m$

---

## ■ 조건

- 노이즈 분포: Gaussian (정규분포)
- 평균: 0
- 표준편차: 0.1

---

## ■ 풀이 절차

### 1️⃣ 실제 값 불러오기
- Practice 1에서 계산한 $r_t$ 사용

---

### 2️⃣ 노이즈 생성

$$
w_t \sim \mathcal{N}(0, 0.1^2)
$$

- 각 시간 step마다 독립적으로 생성

---

### 3️⃣ 측정값 생성

$$
z_t = r_t + w_t
$$

---

### 4️⃣ 결과 시각화

- 실제 값 $r_t$ (ground truth)
- 측정값 $z_t$ (noise 포함)

---

```python
# 시간 배열 생성 
import numpy as np 
import matplotlib.pyplot as plt 

dt = 0.1 
t = np.arange(0,10+dt,dt) # 처음, 끝, 간격 끝은 포함 안됨 

N = len(t)

r=np.zeros(N)
v=np.zeros(N)
a=np.zeros(N)

r[0] = 100
v[0] = 40
a[0] = -9.8

for i in range(N-1): # range(N-1) 이렇게 하면 0부터 N-2 까지 왜냐 r[100] 까지 채워야하는데 i+1 까지 있으니깐 i+1 이 100이 되도록 해야함 그래서 99까지 돌아야 하고 range는 끝 수를 포함 안하니깐 range(100) 이 되도록 range(101)이 되야함 
    r[i+1] = r[i] + v[i]*dt + a[i]*0.5*dt**2
    v[i+1] = v[i] + a[i]*dt
    a[i+1] = a[i]
    
### plot 코드를 줄일 수 있는 방법 ### 


fig, axes = plt.subplots(3,1, figsize=(10,8))

data = [r, v, a]
label = ["altitude", "velocity", "accleration"]

for i in range(3):
    axes[i].plot(t,data[i])
    axes[i].set_ylabel(label[i])
    axes[i].grid()

axes[0].set_title("free fall ")
axes[2].set_xlabel("time")

plt.tight_layout()
plt.show()

```

```python
# 시간 배열 생성 
import numpy as np 
import matplotlib.pyplot as plt 

dt = 0.1 
t = np.arange(0,10+dt,dt)

N = len(t)

r = np.zeros(N)
v = np.zeros(N)
a = np.zeros(N)

r[0] = 100
v[0] = 40
a[0] = -9.8

for i in range(N-1):
    r[i+1] = r[i] + v[i]*dt + 0.5*a[i]*dt**2
    v[i+1] = v[i] + a[i]*dt
    a[i+1] = a[i]
    
fig, axes = plt.subplots(3,1, figsize=(10,8))

data = [r,v,a]
label = ["altitude", "velocity", "accleration"]

for i in range(3):
    axes[i].plot(t,data[i])
    axes[i].set_ylabel(label[i])
    axes[i].grid()
    
axes[0].set_title("free fall")
axes[2].set_xlabel("time")

plt.tight_layout()
plt.show()


```
