# 1D 벡터 행벡터 열벡터 

```python
import numpy as np 

# 1D 배열 - 방향 없음
a = np.array([1, 2, 3])
print(a.shape)        # (3,)

# 행벡터 - 2D, 1행 3열
row = np.array([[1, 2, 3]])
print(row.shape)      # (1, 3)

# 열벡터 - 2D, 3행 1열
col = np.array([[1], [2], [3]])
print(col.shape)      # (3, 1)
```
